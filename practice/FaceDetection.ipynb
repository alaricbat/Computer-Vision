```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

### Import Image

```python
img = cv2.imread('../dataset/img/face.jpg')
```

```python
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
```

### Process Image

```python
B, G, R = cv2.split(img)
zeros = np.zeros(img.shape[:2], dtype=np.uint8)
```

```python
G_img = cv2.merge([zeros, G, zeros])
plt.imshow(G_img)
```

```python
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
```

```python
plt.imshow(hsv)
```

```python
lower = np.array([0, 30, 60])
upper = np.array([20, 170, 255])
```

### Create Mask

```python
brightness_mask = cv2.inRange(hsv[:, :, 2], 80, 255)
```

```python
mask = cv2.inRange(hsv, lower, upper)
```

```python
final_mask = cv2.bitwise_and(mask, brightness_mask)
```

### Get green face

```python
G_img[final_mask == 0] = [255, 255, 255]
```

```python
plt.imshow(G_img)
```
